  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">1. IMPORT LIBRARY</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat pustaka komputasi numerik, manipulasi data, visualisasi grafik, dan kerangka kerja PyTorch.</p>

```python
#202432097_KmsMRizcoAlsyahbiran
import re
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader
from collections import Counter
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Seed agar hasil bisa diulang
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print('Device:', device)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">2. PRA-PEMROSESAN DATA (PREPROCESSING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat dataset, membersihkan data, melakukan normalisasi / tokenisasi sekuens, serta membagi data latih dan data uji.</p>
</div>

```python
#202432097_KmsMRizcoAlsyahbiran
# 1. Muat dataset
DATA_URL = 'https://raw.githubusercontent.com/Alsyahbiran/UTS_DEEP_LEARNING/refs/heads/main/brimo.csv'

if 'USERNAME_GITHUB' in DATA_URL:
    df = pd.read_csv('brimo.csv')      # fallback: file lokal
else:
    df = pd.read_csv(DATA_URL)

print('Jumlah baris & kolom:', df.shape)
print('Nilai kosong per kolom:')
print(df.isna().sum())
df.head()
```

```python
#202432097_KmsMRizcoAlsyahbiran
# 2. Eksplorasi: distribusi rating (score)
score_counts = df['score'].value_counts().sort_index()
print(score_counts)

plt.figure(figsize=(7, 4))
bars = plt.bar(score_counts.index.astype(str), score_counts.values, color='#114F6A')
plt.bar_label(bars, fmt='%d')
plt.title('Distribusi Rating Ulasan BRImo')
plt.xlabel('Score (bintang)')
plt.ylabel('Jumlah ulasan')
plt.grid(True, axis='y', linestyle='--', alpha=0.5)
plt.show()
```

```python
#202432097_KmsMRizcoAlsyahbiran
# 3. Pelabelan sentimen dari score, pembersihan teks, dan penyeimbangan kelas
# score 4-5 -> positif (1), score 1-2 -> negatif (0), score 3 dibuang
df = df[df['score'] != 3].copy()
df['label'] = (df['score'] >= 4).astype(int)

def clean_text(text):
    text = str(text).lower()                      # case folding
    text = re.sub(r'http\S+|www\S+', ' ', text)    # hapus url
    text = re.sub(r'[^a-z\s]', ' ', text)          # hanya huruf (emoji & angka dibuang)
    text = re.sub(r'(.)\1{2,}', r'\1', text)       # huruf berulang: bagusss -> bagus
    text = re.sub(r'\s+', ' ', text).strip()        # rapikan spasi
    return text

df['clean'] = df['content'].apply(clean_text)
df = df[df['clean'].str.len() > 0]                # buang ulasan kosong (mis. hanya emoji)
df = df.drop_duplicates(subset=['clean'])         # buang duplikat agar tidak bocor ke data uji

print('Distribusi label sebelum diseimbangkan:')
print(df['label'].value_counts().rename({1: 'positif', 0: 'negatif'}))

# Ambil sampel seimbang untuk kedua kelas
n_per_class = min(8000, df['label'].value_counts().min())
df_bal = pd.concat([
    df[df['label'] == 1].sample(n_per_class, random_state=SEED),
    df[df['label'] == 0].sample(n_per_class, random_state=SEED)
]).sample(frac=1, random_state=SEED).reset_index(drop=True)

print('\nSampel seimbang per kelas:', n_per_class)
print('Total data dipakai:', len(df_bal))
df_bal[['content', 'clean', 'label']].head()
```

```python
#202432097_KmsMRizcoAlsyahbiran
# 4. Tokenisasi, vocabulary mapping, padding, dan pembagian data
texts = df_bal['clean'].tolist()
labels = df_bal['label'].values

# Bagi data latih 80% dan data uji 20% (stratified)
X_train_txt, X_test_txt, y_train_np, y_test_np = train_test_split(
    texts, labels, test_size=0.2, random_state=SEED, stratify=labels)

# Vocabulary hanya dari data latih
counter = Counter(w for t in X_train_txt for w in t.split())
MAX_VOCAB = 5000
vocab = {'<PAD>': 0, '<UNK>': 1}
for word, freq in counter.most_common(MAX_VOCAB - 2):
    if freq >= 2:
        vocab[word] = len(vocab)
print('Ukuran vocabulary:', len(vocab))

MAX_LEN = 30
def encode(t):
    ids = [vocab.get(w, 1) for w in t.split()][:MAX_LEN]   # kata -> angka, potong jika > MAX_LEN
    return ids + [0] * (MAX_LEN - len(ids))                # padding dengan 0

X_train = torch.tensor([encode(t) for t in X_train_txt], dtype=torch.long)
X_test = torch.tensor([encode(t) for t in X_test_txt], dtype=torch.long)
y_train = torch.tensor(y_train_np, dtype=torch.float32)
y_test = torch.tensor(y_test_np, dtype=torch.float32)

print('X_train:', X_train.shape, '| X_test:', X_test.shape)
print('Contoh teks :', X_train_txt[0])
print('Contoh token:', X_train[0].tolist())

train_loader = DataLoader(TensorDataset(X_train, y_train), batch_size=64, shuffle=True)
test_loader = DataLoader(TensorDataset(X_test, y_test), batch_size=64, shuffle=False)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">3. PEMODELAN ARSITEKTUR JARINGAN (RNN / LSTM)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Mendefinisikan kelas arsitektur model sekuensial menggunakan modul torch.nn sesuai algoritma yang dipilih.</p>
</div>

```python
#202432097_KmsMRizcoAlsyahbiran
class SequenceModel(nn.Module):
    def __init__(self, vocab_size, embed_dim=64, hidden_size=64, dropout=0.3):
        super(SequenceModel, self).__init__()
        self.embedding = nn.Embedding(vocab_size, embed_dim, padding_idx=0)   # angka token -> vektor
        self.lstm = nn.LSTM(embed_dim, hidden_size, batch_first=True)         # layer LSTM
        self.dropout = nn.Dropout(dropout)                                    # regularisasi
        self.fc = nn.Linear(hidden_size, 1)                                   # 1 output (positif/negatif)
        self.sigmoid = nn.Sigmoid()                                           # skor 0 sampai 1

    def forward(self, x):
        emb = self.embedding(x)               # (batch, seq_len, embed_dim)
        _, (h_n, _) = self.lstm(emb)          # h_n: hidden state langkah terakhir
        out = self.dropout(h_n[-1])           # (batch, hidden_size)
        out = self.fc(out)                    # (batch, 1)
        return self.sigmoid(out).squeeze(1)   # (batch,)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">4. INISIASI HYPERPARAMETER & PROSES PELATIHAN (TRAINING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menentukan konfigurasi hyperparameter (learning rate, epochs, loss function, optimizer) dan menjalankan siklus pelatihan model.</p>
</div>

```python
#202432097_KmsMRizcoAlsyahbiran
# 1. Inisiasi hyperparameter
epochs = 15
learning_rate = 0.001

# 2. Inisiasi model, loss function, dan optimizer
model = SequenceModel(vocab_size=len(vocab)).to(device)
criterion = nn.BCELoss()
optimizer = optim.Adam(model.parameters(), lr=learning_rate)
print(model)

# 3. Simpan riwayat untuk visualisasi
train_losses, test_losses, train_accs, test_accs = [], [], [], []

def evaluate(loader):
    model.eval()
    total_loss, correct, n = 0.0, 0, 0
    with torch.no_grad():
        for xb, yb in loader:
            xb, yb = xb.to(device), yb.to(device)
            pred = model(xb)
            total_loss += criterion(pred, yb).item() * len(yb)
            correct += ((pred >= 0.5).float() == yb).sum().item()
            n += len(yb)
    return total_loss / n, correct / n

# 4. Training loop
for epoch in range(epochs):
    model.train()
    run_loss, run_correct, n = 0.0, 0, 0
    for xb, yb in train_loader:
        xb, yb = xb.to(device), yb.to(device)
        optimizer.zero_grad()
        pred = model(xb)
        loss = criterion(pred, yb)
        loss.backward()
        optimizer.step()
        run_loss += loss.item() * len(yb)
        run_correct += ((pred >= 0.5).float() == yb).sum().item()
        n += len(yb)
    tr_loss, tr_acc = run_loss / n, run_correct / n
    te_loss, te_acc = evaluate(test_loader)
    train_losses.append(tr_loss); train_accs.append(tr_acc)
    test_losses.append(te_loss); test_accs.append(te_acc)
    print(f'Epoch {epoch+1:02d}/{epochs} | Train Loss: {tr_loss:.4f} | Train Acc: {tr_acc:.4f} | Test Loss: {te_loss:.4f} | Test Acc: {te_acc:.4f}')
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">5. VISUALISASI LOSS TRAINING & PERBANDINGAN PREDIKSI VS AKTUAL</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menampilkan grafik konvergensi nilai Loss terhadap Epoch dan grafik evaluasi perbandingan hasil prediksi model terhadap data aktual.</p>
</div>

```python
#202432097_KmsMRizcoAlsyahbiran
# 1. Plot grafik penurunan loss dan akurasi terhadap epoch
fig, ax = plt.subplots(1, 2, figsize=(13, 4))
ax[0].plot(train_losses, color='#114F6A', linewidth=2, label='Training Loss')
ax[0].plot(test_losses, color='#1FA4B1', linewidth=2, linestyle='--', label='Test Loss')
ax[0].set_title('Grafik Penurunan Loss Selama Pelatihan')
ax[0].set_xlabel('Epoch'); ax[0].set_ylabel('Loss')
ax[0].grid(True, linestyle='--', alpha=0.5); ax[0].legend()

ax[1].plot(train_accs, color='#114F6A', linewidth=2, label='Training Accuracy')
ax[1].plot(test_accs, color='#1FA4B1', linewidth=2, linestyle='--', label='Test Accuracy')
ax[1].set_title('Grafik Akurasi Selama Pelatihan')
ax[1].set_xlabel('Epoch'); ax[1].set_ylabel('Accuracy')
ax[1].grid(True, linestyle='--', alpha=0.5); ax[1].legend()
plt.tight_layout()
plt.show()
```

```python
#202432097_KmsMRizcoAlsyahbiran
# 2. Evaluasi pada data uji: skor prediksi, classification report, dan confusion matrix
model.eval()
with torch.no_grad():
    scores = model(X_test.to(device)).cpu().numpy()      # skor 0 sampai 1
y_pred = (scores >= 0.5).astype(int)

print('Akurasi data uji:', round(accuracy_score(y_test_np, y_pred), 4))
print()
print(classification_report(y_test_np, y_pred, target_names=['Negatif', 'Positif']))

cm = confusion_matrix(y_test_np, y_pred)
plt.figure(figsize=(5, 4))
plt.imshow(cm, cmap='Blues')
plt.title('Confusion Matrix Data Uji')
plt.xticks([0, 1], ['Negatif', 'Positif']); plt.yticks([0, 1], ['Negatif', 'Positif'])
plt.xlabel('Prediksi'); plt.ylabel('Aktual')
for i in range(2):
    for j in range(2):
        plt.text(j, i, cm[i, j], ha='center', va='center', color='white' if cm[i, j] > cm.max()/2 else 'black', fontsize=13)
plt.colorbar()
plt.show()
```

```python
#202432097_KmsMRizcoAlsyahbiran
# 3. Visualisasi distribusi hasil prediksi sentimen beserta skornya
fig, ax = plt.subplots(1, 2, figsize=(13, 4))

# a. Jumlah prediksi vs jumlah aktual
labels_txt = ['Negatif', 'Positif']
actual_counts = [np.sum(y_test_np == 0), np.sum(y_test_np == 1)]
pred_counts = [np.sum(y_pred == 0), np.sum(y_pred == 1)]
x = np.arange(2)
ax[0].bar(x - 0.2, actual_counts, 0.4, color='#114F6A', label='Aktual')
ax[0].bar(x + 0.2, pred_counts, 0.4, color='#1FA4B1', label='Prediksi')
ax[0].set_xticks(x); ax[0].set_xticklabels(labels_txt)
ax[0].set_title('Distribusi Sentimen: Aktual vs Prediksi')
ax[0].set_ylabel('Jumlah ulasan'); ax[0].set_ylim(0, max(actual_counts + pred_counts) * 1.25); ax[0].legend(loc='upper right')
ax[0].grid(True, axis='y', linestyle='--', alpha=0.5)

# b. Sebaran skor prediksi berdasarkan kelas aktual
ax[1].hist(scores[y_test_np == 0], bins=20, alpha=0.8, color='#114F6A', label='Aktual Negatif')
ax[1].hist(scores[y_test_np == 1], bins=20, alpha=0.8, color='#1FA4B1', label='Aktual Positif')
ax[1].axvline(0.5, color='red', linestyle='--', label='Threshold 0.5')
ax[1].set_title('Sebaran Skor Prediksi (0 = negatif, 1 = positif)')
ax[1].set_xlabel('Skor prediksi'); ax[1].set_ylabel('Jumlah ulasan'); ax[1].legend(loc='upper center')
plt.tight_layout()
plt.show()
```
